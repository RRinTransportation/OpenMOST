```python
import pandas as pd
import json
import os
```

```python
full_fla = pd.read_csv("fla_csvs/code_data.csv")
```

```python
folder = "is-quantitative-p1-gemini"
results = {}

for filename in os.listdir(folder):
    if filename.endswith(".json"):
        file_path = os.path.join(folder, filename)
        with open(file_path, "r", encoding="utf-8") as f:
            data = json.load(f)
            is_quantitative = data.get("is_quantitative_study")
            results[filename.replace("_", "/").strip(".json")] = is_quantitative
```

```python
full_fla["is_quantitative_study"] = full_fla["doi"].map(results)
```
