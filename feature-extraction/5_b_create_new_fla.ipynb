```python
import pandas as pd
import json
```

```python
fla = pd.read_csv("fla_csvs/meta.csv")

def add_to_fla(folder, features):
    for index, row in fla.iterrows():
        with open(f"{folder}/{row['doi'].replace('/', '_')}.json") as f:
            full_llm_code = json.load(f)

        for feature in features:
            if feature in full_llm_code:
                fla.loc[index, feature] = str(full_llm_code[feature])
```

```python
fla["is_code_used"] = None
fla["is_code_publicly_available"] = None
fla["code_link"] = None
fla["code_link_by_the_author"] = None
fla["code_link"] = None
fla["programming_language"] = None
fla["software_used"] = None
fla["cpu_used"] = None
fla["gpu_used"] = None
fla["os_used"] = None
fla["ram_used"] = None

fla["is_data_used"] = None
fla["is_simulation_study"] = None
fla["is_data_repository_available"] = None
fla["links_to_the_data_repository"] = None
fla["is_data_cited_or_linked"] = None
fla["links_to_the_data_cited_or_linked"] = None
```

```python
code_features = [
    "is_code_used", 
    "is_code_publicly_available",
    "code_link", 
    "code_link_by_the_author",
    "code_link",
    "programming_language",
    "software_used",
    "cpu_used",
    "gpu_used",
    "os_used",
    "ram_used"
]

add_to_fla("code-p1-gemini", code_features)
```

```python
data_features = [
    "is_data_used",
    "is_simulation_study",
    "is_data_repository_available",
    "links_to_the_data_repository",
    "is_data_cited_or_linked",
    "links_to_the_data_cited_or_linked"
]

add_to_fla("data-p1-gemini", data_features)
```

```python
fla.to_csv("fla_csvs/code_data.csv")
```
